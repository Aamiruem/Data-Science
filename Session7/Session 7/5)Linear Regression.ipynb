```python
# Importing the function make_regression from sklearn.datasets
# This function is used to create a fake (synthetic) regression dataset for testing or practice.
from sklearn.datasets import make_regression

# Creating input data (x) and target output (y)
# n_features=1     → each sample will have 1 feature (i.e., a single x value)
# noise=5          → adds random noise to make the data more realistic (like real-world data)
# n_samples=10000  → generate 10,000 data points (rows)
# random_state=1   → ensures the same random data is generated every time (for reproducibility)
x, y = make_regression(n_features=1, noise=5, n_samples=10000, random_state=1)

# Importing the matplotlib.pyplot library and naming it as plt
# This library is used for plotting graphs
import matplotlib.pyplot as plt

# Creating a scatter plot (dots) of x (input features) vs y (target values)
plt.scatter(x, y)

# Labeling the x-axis as "x"
plt.xlabel('x')

# Labeling the y-axis as "y"
plt.ylabel('y')

# Displaying the plot window with the graph
plt.show()
```

```python
from sklearn.model_selection import train_test_split 
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state=0)
```

```python
# Import the train_test_split function from scikit-learn's model_selection module
# This function is used to split your dataset into training and testing subsets
from sklearn.model_selection import train_test_split

# Use train_test_split to randomly divide the data into training and test sets
# The function returns four variables:
# X_train → the training input data (features)
# X_test  → the testing input data (features)
# y_train → the training output data (labels/target values)
# y_test  → the testing output data (labels/target values)
X_train, X_test, y_train, y_test = train_test_split(

    x,  # Input features (independent variables) — usually a 2D array or matrix
    y,  # Target/output values (dependent variable) — usually a 1D array or list

    test_size=0.2,  # This means 20% of the data will be used for testing, and 80% for training

    random_state=0  # This is the seed for random number generator; it ensures reproducibility
    # Setting random_state to the same value ensures you get the same split every time you run the code
)


# from sklearn.model_selection	Importing from sklearn's submodule that handles data splitting
# train_test_split	Function used to divide data into training and testing sets
# x	Input/features (independent variables)
# y	Output/labels (dependent variable)
# test_size=0.2	20% of the dataset goes to testing; the rest to training
# random_state=0	Sets a fixed seed to ensure the same split every time
# X_train	80% of x (features) for training
# X_test	20% of x (features) for testing
# y_train	80% of y (labels) for training
# y_test	20% of y (labels) for testing
```

```python
from sklearn.linear_model import LinearRegression 
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
# Import the LinearRegression class from sklearn's linear_model module
# This class is used to create a linear regression model
from sklearn.linear_model import LinearRegression

# Create an instance (object) of the LinearRegression model and assign it to the variable 'regressor'
# This object will be used to train (fit) and make predictions
regressor = LinearRegression()

# Train the linear regression model using the training data (X_train and y_train)
# 'fit()' means the model learns the relationship between the input features and the output labels
# Specifically, it finds the best-fit line (y = mx + c) that minimizes error
regressor.fit(X_train, y_train)



# from	Python keyword used to import a part of a module
# sklearn	Scikit-learn, a popular Python library for machine learning
# linear_model	Submodule inside sklearn that contains models for linear algorithms
# LinearRegression	A class in sklearn used to create a simple or multiple linear regression model
# regressor	A variable name to hold the created model (you can name it anything, like model)
# LinearRegression()	Calls the class constructor to create a new linear regression model
# fit()	A method that trains the model using the input features (X_train) and corresponding output (y_train)

# 🔍 What does fit() do?
# The fit(X_train, y_train) function:

# Learns the pattern or relationship between X_train (independent variable) and y_train (dependent variable)

# For linear regression, it calculates the best line of fit by finding the optimal values for:

# slope(s) (also called coefficients)

# intercept (also called bias)

# After fitting, the model can now predict values using .predict()

# 💡 Real-world Analogy:
# Think of .fit() like teaching the model. You're showing it examples of:

# inputs: "square feet of a house"

# outputs: "house price"
# And it learns how house size affects the price, so it can predict prices later on.
```

```python
# Use the trained regression model (regressor) to make predictions on new, unseen input data (X_test)

# 'regressor' is the object of LinearRegression that was trained using .fit(X_train, y_train)
# It already learned the relationship between inputs and outputs during training

# 'predict()' is a method that uses the learned model (the best-fit line equation: y = mx + c)
# to calculate and return the predicted output values (y) for the given input data (X_test)

# 'X_test' is the test set of input features (independent variables) that the model has never seen before

# This function returns a list or array of predicted values corresponding to each row in X_test
# These can later be compared to the actual y_test values to evaluate the model's accuracy
regressor.predict(X_test)
```
