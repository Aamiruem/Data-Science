# lambda function is anonymous(unnamed) function 

```python
numbers = [1, 2, 3, 4, 5, 6]

result = map(lambda x: x * 2, numbers)

print(list(result))
```

```python
numbers = [1, 2, 3, 4, 5, 6]

result = filter(lambda x: x % 2 == 0, numbers)

print(list(result))
```

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5, 6, 7]

result = reduce(lambda a, b: a + b, numbers)

print(result)
```

```python
def countdown(n):
    if n == 0:
        return
    print(n)

    countdown(n - 1)


countdown(5)
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

```python
def decorator(func):
    def wrapper():
        func()

    return wrapper


@decorator
def greet():
    print("Hello")


greet()
```

```python
def my_decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")

    return wrapper


@my_decorator
def greet():
    print("Hello")


greet()
```

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Before")
        result = func(*args, **kwargs)
        print("After")
        return result

    return wrapper
```

```python
def logger(func):
    def wrapper(*args, **kwargs):
        print("Function started")
        result = func(*args, **kwargs)
        print("Function finished")
        return result

    return wrapper


@logger
def add(a, b):
    return a + b


print(add(10, 20))
```

```python
from functools import wraps

def decorator(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    
    return wrapper   
```
