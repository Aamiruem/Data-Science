```python
def factorial(n):
    if n < 0:
        raise ValueError("N must be a non-negative integer.")
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(f"The factorial of 5 is: {factorial(5)}")
```

```python
def normalize(value: float, minimum: float, maximum: float) -> float:
    if maximum <= minimum:
        raise ValueError("Maximum must be greater than minimum.")
    return (value - minimum) / (maximum - minimum)
print(f"The normalized value is: {normalize(75, 0, 100)}")
```

```python
from sklearn.metrics import mean_absolute_error


def prepare_features(study_hours):
    if study_hours < 0:
        raise ValueError("Study hours cannot be Negative.")

# 2. class for prediction model
class ScorePredictor:
    def __init__(self, rate = 10):
        self.rate = rate

    def predict(self, study_hours):
        hours = prepare_features(study_hours)
        return min(hours * self.rate, 100)

    # 3. Evaluation function 
def mean_absolute_error(actual, predicted):
        if len(actual) != len(predicted) or not actual:
            raise ValueError("Lists must have equal, nonzero length")

        errors = [
            abs(a - p)
            for a, p in zip(actual, predicted)
            
        ]
        return sum(errors)/len(errors)
#  4. use the system
model = ScorePredictor(rate = 10)

hours = [2, 5, 8]
def prepare_features(study_hours):
    if study_hours < 0:
        raise ValueError("Study hours cannot be negative.")
    return study_hours

actual_scores = [25, 55, 75]
predictions = [model.predict(hour) for hour in hours]
mae = mean_absolute_error(actual_scores, predictions)

print(f"Predictions: {predictions}")
print(f"Mean Absolute Error: {mae}")
print("MAE:", mae)
```

```python
class BankAccount:
    def __init__(self, balance = 0):
        if balance < 0:
            raise ValueError("Initial balance cannot be Negative")
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be Positive")
        self._balance += amount


    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal must be Positive")
        if amount > self._balance:
            raise ValueError("Insufficient funds")
        self._balance -= amount



account = BankAccount(100)
print(f"Initial balance: ${account.balance}")

account.deposit(50)
print(f"Balance after deposit: ${account.balance}")
print(account.balance)


withdraw_amount = 30
account.withdraw(withdraw_amount)
print(f"Withdraw Amount: ${withdraw_amount}, Balance after withdrawal: ${account.balance}")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler


# Function: check the input data
def validate_features(X):
    if len(X) == 0:
        raise ValueError("Training data cannot be empty")
    if any(len(row) != 1 for row in X):
        raise ValueError("Each row must contain one feature")
    return X


# Class: organize the training and prediction operations
class HousePriceModel:
    def __init__(self):
        self.pipeline = Pipeline([
            ("scaler", StandardScaler()),
            ("model", LinearRegression())
        ])

    def train(self, X, y):
        X = validate_features(X)

        if len(X) != len(y):
            raise ValueError("X and y must have equal lengths")

        self.pipeline.fit(X, y)
        return self

    def predict(self, X):
        X = validate_features(X)
        return self.pipeline.predict(X)


# Demonstration dataset: house size in square feet
X_train = [[500], [700], [900], [1100], [1300]]
y_train = [100, 140, 180, 220, 260]  # Example price units

X_test = [[800], [1000]]
y_test = [160, 200]

# Create an object
model = HousePriceModel()

# Train the model
model.train(X_train, y_train)

# Predict on held-out demonstration samples
predictions = model.predict(X_test)

# Evaluate prediction errors
mae = mean_absolute_error(y_test, predictions)

print("Predictions:", predictions)
print("Actual values:", y_test)
print("MAE:", mae)
```

```python
# Service class that uses a model
class PredictionService:

    # Constructor
    def __init__(self, model):
        # Store model object
        self.model = model

    # Predict method
    def predict(self, features):
        # Call model's predict method
        return self.model.predict(features)


# Demo model class
class DemoModel:

    # Predict method
    def predict(self, features):
        # Return sum of all feature values
        return sum(features)


# Create service with DemoModel
service = PredictionService(DemoModel())

# Make prediction
print(service.predict([10, 20, 30]))
```

```python
def calculate_average(scores):
    if len(scores) == 0:
        return None

    total = sum(scores)

    count = len(scores)
    average = total / count
    return average
model_score = [80, 90, 100]

result  = calculate_average(model_score)
print("Average score:", result)
```
