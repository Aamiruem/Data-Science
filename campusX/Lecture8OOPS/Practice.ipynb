```python
def factorial(n):
    if n < 0:
        raise ValueError("N must be a non-negative integer.")
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(f"The factorial of 5 is: {factorial(5)}")
```

```python
def normalize(value: float, minimum: float, maximum: float) -> float:
    if maximum <= minimum:
        raise ValueError("Maximum must be greater than minimum.")
    return (value - minimum) / (maximum - minimum)
print(f"The normalized value is: {normalize(75, 0, 100)}")
```

```python
from sklearn.metrics import mean_absolute_error


def prepare_features(study_hours):
    if study_hours < 0:
        raise ValueError("Study hours cannot be Negative.")

# 2. class for prediction model
class ScorePredictor:
    def __init__(self, rate = 10):
        self.rate = rate

    def predict(self, study_hours):
        hours = prepare_features(study_hours)
        return min(hours * self.rate, 100)

    # 3. Evaluation function 
def mean_absolute_error(actual, predicted):
        if len(actual) != len(predicted) or not actual:
            raise ValueError("Lists must have equal, nonzero length")

        errors = [
            abs(a - p)
            for a, p in zip(actual, predicted)
            
        ]
        return sum(errors)/len(errors)
#  4. use the system
model = ScorePredictor(rate = 10)

hours = [2, 5, 8]
def prepare_features(study_hours):
    if study_hours < 0:
        raise ValueError("Study hours cannot be negative.")
    return study_hours

actual_scores = [25, 55, 75]
predictions = [model.predict(hour) for hour in hours]
mae = mean_absolute_error(actual_scores, predictions)

print(f"Predictions: {predictions}")
print(f"Mean Absolute Error: {mae}")
print("MAE:", mae)
```

```python
class BankAccount:
    def __init__(self, balance = 0):
        if balance < 0:
            raise ValueError("Initial balance cannot be Negative")
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be Positive")
        self._balance += amount


    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal must be Positive")
        if amount > self._balance:
            raise ValueError("Insufficient funds")
        self._balance -= amount



account = BankAccount(100)
print(f"Initial balance: ${account.balance}")

account.deposit(50)
print(f"Balance after deposit: ${account.balance}")
print(account.balance)


withdraw_amount = 30
account.withdraw(withdraw_amount)
print(f"Withdraw Amount: ${withdraw_amount}, Balance after withdrawal: ${account.balance}")
```

### Methods Overriding

```python
class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def speak(self):
        print("Dog barks")
        super().speak()

class Cat(Animal):
    def speak(self):
        print("Cat meows")
        super().speak()

dog = Dog()
cat = Cat()
dog.speak()
cat.speak()
```

### Duck Typing

```python
class Dog:
    def speak(self):
        print("Dog barks")
class Cat:
    def speak(self):
        print("Mewo")

def make_sound(animal):
    animal.speak()

make_sound(Dog())
make_sound(Cat()) 
```

```python
class Employee:
    def calculate_salary(self):
        pass # This method will be overridden by subclasses 

class FullTimeEmployee(Employee):
    def calculate_salary(self):
        return f"Full-time salary: $50,000"

class PartTimeEmployee(Employee):
    def calculate_salary(self):
        return f"Part-time salary: $30,000"

employees = [FullTimeEmployee(), PartTimeEmployee()]
for employee in employees:
    print(employee.calculate_salary())
```

```python
from abc import ABC, abstractmethod

class Predictor(ABC):
    @abstractmethod
    def predict(self, features):
        pass

class SimplePredictor(Predictor):
    def predict(self, features):
        return sum(features) / len(features)

model = SimplePredictor()
print(model.predict([10, 20, 30]))
```
