```python
number = int(input("enter the number"))

digits = "0123456789"
result = ""
while number != 0:
    result = digits[number % 10] + result
    number = number // 10

print(result)
```

```python
L = [1, 2, 3, 4]

sum = 0
for i in L:
    sum = sum + i

product = 1
for i in L:
    product = product * i

print(sum, product)
```

```python
L = [1, 2, 3, 4, 5]

for i in range(0, len(L) // 2):
    other = len(L) - i - 1
    temp = L[i]
    L[i] = L[other]
    L[other] = temp

print(L)
```

```python
n = 10
k = 0
for i in range(n // 2, n):
    for j in range(2, n, pow(2, j)):
        k = k + n / 2

print(k)
```

```python
a = 10
b = 3

if b <= 0:
    print(-1)
div = a // b

print(a - div - b)
```

```python
n = 345

sum = 0
while n > 0:
    sum = sum + n % 10
    n = n // 10

print(sum)
```

```python
def fib(n):
    if n == 1 or n == 0:
        return 1
    else:
        return fib(n - 1) + fib(n - 2)
```
