```python

import pygame
import random
import math
import sys

# ==========================================
# HELICOPTER FLYING SIMULATOR
# ==========================================

pygame.init()

WIDTH, HEIGHT = 1200, 700
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("🚁 Advanced Helicopter Flying Simulator")

clock = pygame.time.Clock()
FPS = 60

# Colors
WHITE = (255, 255, 255)
BLACK = (15, 20, 35)
SKY = (80, 190, 255)
GREEN = (30, 180, 90)
RED = (255, 60, 60)
YELLOW = (255, 220, 40)
BLUE = (30, 100, 255)
ORANGE = (255, 130, 20)
PURPLE = (160, 80, 255)

font = pygame.font.SysFont("arial", 25, bold=True)
big_font = pygame.font.SysFont("arial", 55, bold=True)

# Player helicopter
heli = pygame.Rect(160, 300, 100, 50)
heli_y = float(heli.y)

velocity_y = 0
gravity = 0.25
lift = -0.55
max_speed = 6

fuel = 100
health = 100
score = 0
distance = 0

rotor_angle = 0
camera_x = 0

game_over = False
paused = False
started = False

# Clouds
clouds = [
    [random.randint(0, WIDTH), random.randint(30, 220),
     random.randint(30, 65)]
    for _ in range(12)
]

# Mountains
mountains = [
    [i * 300, random.randint(390, 510)]
    for i in range(12)
]

# Buildings / obstacles
obstacles = []

for i in range(10):
    obstacles.append(
        pygame.Rect(
            600 + i * 400,
            random.randint(420, 540),
            random.randint(55, 100),
            random.randint(100, 220)
        )
    )

# Landing pad
landing_pad = pygame.Rect(1800, 560, 160, 20)

# Particles
particles = []

def draw_text(text, x, y, color=WHITE, font_size=None):
    selected_font = font if font_size is None else pygame.font.SysFont(
        "arial", font_size, bold=True
    )
    surface = selected_font.render(str(text), True, color)
    screen.blit(surface, (x, y))


def draw_cloud(x, y, size):
    pygame.draw.circle(screen, WHITE, (int(x), int(y)), size)
    pygame.draw.circle(
        screen, WHITE, (int(x + size), int(y - 10)), size
    )
    pygame.draw.circle(
        screen, WHITE, (int(x + size * 2), int(y)), size
    )
    pygame.draw.ellipse(
        screen, WHITE,
        (x, y, size * 2, size)
    )


def draw_helicopter(x, y):
    global rotor_angle

    # Shadow
    pygame.draw.ellipse(
        screen, (30, 40, 50),
        (x + 10, y + 48, 90, 12)
    )

    # Tail
    pygame.draw.polygon(
        screen, ORANGE,
        [(x + 15, y + 22),
         (x - 35, y + 35),
         (x - 35, y + 41),
         (x + 20, y + 37)]
    )

    # Tail rotor
    pygame.draw.line(
        screen, WHITE,
        (x - 35, y + 27),
        (x - 35, y + 48), 4
    )
    pygame.draw.line(
        screen, WHITE,
        (x - 45, y + 37),
        (x - 25, y + 37), 4
    )

    # Main body
    pygame.draw.ellipse(
        screen, GREEN,
        (x + 5, y + 15, 85, 38)
    )

    # Cockpit
    pygame.draw.ellipse(
        screen, (80, 210, 255),
        (x + 55, y + 18, 30, 24)
    )

    # Cockpit reflection
    pygame.draw.line(
        screen, WHITE,
        (x + 70, y + 20),
        (x + 80, y + 27), 2
    )

    # Landing skids
    pygame.draw.line(
        screen, BLACK,
        (x + 25, y + 52),
        (x + 85, y + 52), 4
    )

    pygame.draw.line(
        screen, BLACK,
        (x + 35, y + 47),
        (x + 35, y + 53), 3
    )

    pygame.draw.line(
        screen, BLACK,
        (x + 75, y + 47),
        (x + 75, y + 53), 3
    )

    # Rotor mast
    pygame.draw.line(
        screen, BLACK,
        (x + 45, y + 17),
        (x + 45, y + 5), 5
    )

    # Animated main rotor
    rotor_angle += 30

    rad = math.radians(rotor_angle)

    dx = math.cos(rad) * 60
    dy = math.sin(rad) * 8

    pygame.draw.line(
        screen, BLACK,
        (x + 45 - dx, y + 5 - dy),
        (x + 45 + dx, y + 5 + dy), 5
    )


def draw_background():
    # Sky gradient
    for i in range(HEIGHT):
        color = (
            40 + int(i * 0.07),
            150 + int(i * 0.08),
            235 - int(i * 0.05)
        )
        pygame.draw.line(screen, color, (0, i), (WIDTH, i))

    # Sun
    pygame.draw.circle(screen, YELLOW, (1030, 110), 55)

    # Clouds
    for cloud in clouds:
        draw_cloud(cloud[0], cloud[1], cloud[2])

        cloud[0] -= 0.5

        if cloud[0] < -150:
            cloud[0] = WIDTH + random.randint(20, 200)


def draw_terrain():
    pygame.draw.rect(
        screen, (40, 180, 70),
        (0, 560, WIDTH, 140)
    )

    # Ground
    pygame.draw.rect(
        screen, (110, 75, 40),
        (0, 610, WIDTH, 90)
    )

    # Mountains
    for mountain_x, mountain_y in mountains:
        x = mountain_x - camera_x

        if -300 < x < WIDTH + 300:
            pygame.draw.polygon(
                screen, (75, 120, 145),
                [(x, 560),
                 (x + 150, mountain_y),
                 (x + 300, 560)]
            )

            pygame.draw.polygon(
                screen, WHITE,
                [(x + 110, mountain_y + 55),
                 (x + 150, mountain_y),
                 (x + 190, mountain_y + 55)]
            )


def draw_obstacles():
    for obstacle in obstacles:
        x = obstacle.x - camera_x

        if -150 < x < WIDTH + 150:
            pygame.draw.rect(
                screen, (90, 90, 105),
                (x, obstacle.y, obstacle.width, obstacle.height)
            )

            pygame.draw.rect(
                screen, YELLOW,
                (x + 10, obstacle.y + 15, 10, 12)
            )

            pygame.draw.rect(
                screen, YELLOW,
                (x + 35, obstacle.y + 15, 10, 12)
            )


def draw_landing_pad():
    x = landing_pad.x - camera_x

    if -200 < x < WIDTH + 200:
        pygame.draw.rect(
            screen, WHITE,
            (x, landing_pad.y, landing_pad.width, 8)
        )

        draw_text("H", x + 60, landing_pad.y - 45, YELLOW, 40)


def draw_hud():
    pygame.draw.rect(
        screen, (15, 25, 45),
        (15, 15, 350, 150), border_radius=15
    )

    draw_text("🚁 FLIGHT CONTROL", 30, 25, YELLOW)

    draw_text(f"FUEL: {int(fuel)}%", 30, 65)

    pygame.draw.rect(screen, RED, (150, 72, 170, 15))
    pygame.draw.rect(
        screen, GREEN,
        (150, 72, 170 * max(0, fuel) / 100, 15)
    )

    draw_text(f"HEALTH: {int(health)}%", 30, 100)

    pygame.draw.rect(screen, RED, (170, 108, 150, 15))
    pygame.draw.rect(
        screen, BLUE,
        (170, 108, 150 * max(0, health) / 100, 15)
    )

    draw_text(f"SCORE: {score}", 30, 135, YELLOW)


def reset_game():
    global heli_y, velocity_y, fuel, health
    global score, distance, camera_x, game_over
    global started, paused

    heli_y = 300.0
    heli.y = int(heli_y)

    velocity_y = 0
    fuel = 100
    health = 100
    score = 0
    distance = 0
    camera_x = 0

    game_over = False
    started = True
    paused = False


# Main game loop
running = True

while running:

    clock.tick(FPS)

    for event in pygame.event.get():

        if event.type == pygame.QUIT:
            running = False

        if event.type == pygame.KEYDOWN:

            if event.key == pygame.K_ESCAPE:
                running = False

            if event.key == pygame.K_RETURN:
                if not started or game_over:
                    reset_game()

            if event.key == pygame.K_p and started:
                paused = not paused

    keys = pygame.key.get_pressed()

    if started and not game_over and not paused:

        # Flying physics
        if keys[pygame.K_UP] or keys[pygame.K_w]:
            velocity_y += lift

        if keys[pygame.K_DOWN] or keys[pygame.K_s]:
            velocity_y += 0.4

        if keys[pygame.K_LEFT] or keys[pygame.K_a]:
            camera_x = max(0, camera_x - 3)

        if keys[pygame.K_RIGHT] or keys[pygame.K_d]:
            camera_x += 4
            distance += 4

        velocity_y += gravity

        velocity_y = max(-max_speed, min(max_speed, velocity_y))

        heli_y += velocity_y
        heli.y = int(heli_y)

        # Fuel consumption
        fuel -= 0.025

        if keys[pygame.K_UP] or keys[pygame.K_w]:
            fuel -= 0.035

        fuel = max(0, fuel)

        if fuel <= 0:
            velocity_y += 0.1

        # Boundary detection
        if heli.top < 0:
            heli.top = 0
            heli_y = 0
            health -= 0.2

        if heli.bottom > 560:
            heli.bottom = 560
            heli_y = heli.y
            velocity_y = 0
            health -= 0.15

        # Obstacle collision
        player_box = pygame.Rect(
            heli.x - 30, heli.y + 10, 125, 40
        )

        for obstacle in obstacles:

            obstacle_screen = pygame.Rect(
                obstacle.x - camera_x,
                obstacle.y,
                obstacle.width,
                obstacle.height
            )

            if player_box.colliderect(obstacle_screen):
                health -= 1

        # Landing pad refueling
        pad_screen = pygame.Rect(
            landing_pad.x - camera_x,
            landing_pad.y - 15,
            landing_pad.width,
            30
        )

        if player_box.colliderect(pad_screen):
            if abs(velocity_y) < 2:
                fuel = min(100, fuel + 0.5)
                health = min(100, health + 0.2)
                score += 1

        # Score
        score = int(distance // 10)

        if health <= 0 or heli.bottom >= HEIGHT:
            game_over = True

    # Rendering
    draw_background()
    draw_terrain()
    draw_obstacles()
    draw_landing_pad()

    if started:
        draw_helicopter(heli.x, heli.y)

    draw_hud()

    # Start screen
    if not started:

        overlay = pygame.Surface((WIDTH, HEIGHT))
        overlay.set_alpha(180)
        overlay.fill(BLACK)
        screen.blit(overlay, (0, 0))

        title = big_font.render(
            "HELICOPTER SIMULATOR",
            True, YELLOW
        )

        screen.blit(
            title,
            (WIDTH // 2 - title.get_width() // 2, 180)
        )

        draw_text(
            "Press ENTER to start flying",
            430, 290, WHITE, 32
        )

        draw_text(
            "UP / W = Fly Up | DOWN / S = Fly Down",
            365, 370, GREEN
        )

        draw_text(
            "LEFT / RIGHT = Move | P = Pause",
            375, 415, WHITE
        )

    # Pause screen
    if paused:

        draw_text(
            "GAME PAUSED",
            470, 300, YELLOW, 55
        )

        draw_text(
            "Press P to Continue",
            480, 370, WHITE, 30
        )

    # Game over
    if game_over:

        overlay = pygame.Surface((WIDTH, HEIGHT))
        overlay.set_alpha(190)
        overlay.fill(BLACK)
        screen.blit(overlay, (0, 0))

        draw_text(
            "FLIGHT OVER!",
            430, 220, RED, 60
        )

        draw_text(
            f"FINAL SCORE: {score}",
            470, 310, YELLOW, 35
        )

        draw_text(
            "Press ENTER to Restart",
            420, 390, WHITE, 30
        )

    pygame.display.flip()

pygame.quit()
sys.exit()
```

```python
def read_dataset(filename):

    try:

        with open(filename, "r") as file:

            data = file.read()

            if not data.strip():
                raise ValueError("Dataset is empty")

            print("Dataset loaded successfully")
            print(data)

    except FileNotFoundError:

        print("Error: Dataset file not found")

    except PermissionError:

        print("Error: Permission denied")

    except ValueError as e:

        print("Error:", e)

    finally:

        print("Dataset operation completed")


read_dataset("dataset.csv")
```

```python
import json

def save_results(filename, results):

    try:

        with open(filename, "w") as file:

            json.dump(results, file, indent=4)

        print("Results saved successfully")

    except PermissionError:

        print("You don't have permission to write this file")

    except OSError as e:

        print("File operation failed:", e)


results = {
    "model": "Random Forest",
    "accuracy": 0.95,
    "precision": 0.92,
    "recall": 0.94
}

save_results("model_results.json", results)
```

```python
import csv

def validate_dataset(filename):
    try:
        with open(filename, "r", newline="") as file:

            reader = csv.DictReader(file)
            required_columns = ["Name", "Age", "Salary", "Target"]

            # Check if header exists
            if reader.fieldnames is None:
                raise ValueError("CSV header is missing")

            # Check for missing columns
            missing_columns = [
                column
                for column in required_columns
                if column not in reader.fieldnames
            ]

            if missing_columns:
                raise ValueError(
                    f"Missing columns: {missing_columns}"
                )

            print("Dataset validation successful")

    except FileNotFoundError:
        print("Dataset file not found")

    except ValueError as e:
        print("Dataset validation failed:", e)

    except csv.Error as e:
        print("CSV processing error:", e)

validate_dataset("training_data.csv")
```

```python
import re

with open("data.txt", "r") as file:
    content = file.read()

print("=" * 50)
print("CONTENT:")
print("=" * 50)
print(content)
print("=" * 50)

# --- Total characters (everything) ---
total_chars = len(content)

# --- Words (sequences of letters/digits, ignoring punctuation) ---
words = re.findall(r"\b\w+\b", content)
word_count = len(words)

# --- Characters (letters only, A-Z / a-z) ---
letter_count = sum(1 for ch in content if ch.isalpha())

# --- Digits ---
digit_count = sum(1 for ch in content if ch.isdigit())

# --- Spaces (space, tab, newline) ---
space_count = sum(1 for ch in content if ch.isspace())

# --- Sentences (split on . ! ?) ---
sentences = [s for s in re.split(r"[.!?]+", content) if s.strip()]
sentence_count = len(sentences)

# --- Special characters (not letter, not digit, not whitespace) ---
special_count = sum(
    1 for ch in content if not ch.isalnum() and not ch.isspace()
)

# --- Print results ---
print(f"Total characters (all) : {total_chars}")
print(f"Letters (A-Z, a-z)     : {letter_count}")
print(f"Words                  : {word_count}")
print(f"Sentences              : {sentence_count}")
print(f"Spaces (incl. \\n, \\t)  : {space_count}")
print(f"Digits (0-9)           : {digit_count}")
print(f"Special characters     : {special_count}")

# --- Optional: verify the parts add up ---
print("-" * 50)
print(f"Check: letters + digits + spaces + special = "
      f"{letter_count + digit_count + space_count + special_count} "
      f"(should equal total {total_chars})")
```
