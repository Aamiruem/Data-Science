```python
Name = ["Aamir", "Rahul", "Priya"]
Score = [98, 94, 96]

result = [[Name, Score] for Name, Score in zip(Name, Score)]
print(result)
```

```python
result = []
for i in range(len(Name)):
    result.append([Name[i], Score[i]])

print(result)
```

```python
fruits = ["Apple", "Banana", "Grapes", "Pineapple"]

target = "Banana"
new_item = "Mango"

if target in fruits:
    index = fruits.index(target)
    fruits.insert(index + 1, new_item)
print(fruits)
```
