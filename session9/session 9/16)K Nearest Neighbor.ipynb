```python
# Importing the KNeighborsClassifier class from the sklearn.neighbors module
from sklearn.neighbors import KNeighborsClassifier  

# Importing the pyplot module from matplotlib, renaming it as plt 
# (plt is a common alias used for convenience)
import matplotlib.pyplot as plt  

# Importing the make_blobs function from sklearn.datasets
# make_blobs is used to generate sample datasets (clusters of points) for testing ML algorithms
from sklearn.datasets import make_blobs  
```

```python
x,y = make_blobs(n_features=2, n_samples=1000, cluster_std=2, centers=2,random_state=1)

# Generate synthetic dataset with blob clusters
x, y = make_blobs(
    n_features=2,      # Number of features/dimensions for each sample (2D data for easy plotting)
    n_samples=1000,    # Total number of data points to generate
    cluster_std=2,     # Standard deviation of the clusters (higher = more spread out clusters)
    centers=2,         # Number of clusters/centers to generate (binary classification)
    random_state=1     # Seed for random number generator (ensures reproducible results)
)
```

```python
plt.scatter(x[:,0], x[:,1],alpha=0.6,c=y)
plt.scatter(-5,0,c='r')
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size = 0.3, random_state=0)
```

```python
kn = KNeighborsClassifier(n_neighbors=5)
```

```python
kn.fit(X_train,y_train)
```

```python
kn.score(x,y)
```

```python
neighbors=kn.kneighbors([[-5,0]])
neighbors
```

```python
_,neighbors = kn.kneighbors([[-5,0]])
```

```python
x[neighbors[0]]
```

```python
from sklearn.datasets import make_regression
x,y = make_regression(n_features=1,noise=20,n_samples=10000,random_state=1)
import matplotlib.pyplot as plt
plt.scatter(x,y,s=3)
plt.show()
```

```python
from sklearn.neighbors import KNeighborsRegressor
```

```python
kn = KNeighborsRegressor(n_neighbors=3)
```

```python
kn.fit(x,y)
```

```python
kn.score(x,y)
```

```python
_,neighbors = kn.kneighbors([[1]])
```

```python
y[neighbors[0]]
```
