```python
from sklearn.datasets import make_regression

# Generate sample data
X, y = make_regression(n_samples=100, n_features=1, noise=10, random_state=42)

print(X[:5])  # First 5 input values
print(y[:5])  # Corresponding target values
```

```python

# 📘 Parameters:
# n_samples=10000: Number of data points (rows)

# n_features=1: Number of features (columns)

# noise=10: Adds random noise to simulate real-world data

# random_state=0: Ensures the same output every time (for reproducibility)


x,y = make_regression(n_features=1, noise=0.0000000001, n_samples=10000,random_state=0)

# X: 100 rows × 1 column of random input data (generated features)

# y: 100 corresponding target values, linearly dependent on X (with noise)
```

```python
y
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
