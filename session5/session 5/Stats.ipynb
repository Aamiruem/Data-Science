```python
import numpy as np 
x = [10,20,30,30,45,50,55,60,70,80]
```

# Mean
in data science, the mean (or average) is a fundamental measure of central tendency that represents the sum of all values divided by the number of values. It is widely used in statistics, machine learning, and data analysis to understand the central value of a dataset.

# 1. Mathematical Definition
The mean (arithmetic mean) of a dataset is calculated as:

```python
np.mean(x)
```

```python
np.median(x)
```

```python
np.percentile(x, [25,50,75,100])
```

```python
np.std(x)
```

```python
np.var(x)
```
