# Median


📊 What is Median in Data Science?
Median is a measure of central tendency — it represents the middle value in a sorted list of numbers.

🧠 Definition:
The median is the value that separates the higher half from the lower half of a dataset.

For odd number of elements:
Just pick the middle value after sorting.

Example:
Data: [3, 1, 7]
Sorted: [1, 3, 7]
Median = 3
For even number of elements:
Take the average of the two middle values.

Example:
Data: [4, 1, 6, 2]
Sorted: [1, 2, 4, 6]
Median = (2 + 4) / 2 = 3
🔍 Why is Median Useful?
It isn’t affected by outliers as much as the mean (average).

It's great for skewed data.

Example:
Salaries: [30k, 32k, 31k, 33k, 1 million]
Mean ≈ 225k (misleading)
Median = 32k (more accurate)

🧮 Python Example:
import statistics
data = [3, 1, 7]
print(statistics.median(data))  # Output: 3

```python
np.median(x)
```

# Standard Deviation

📊 Standard Deviation — Short & Clear Explanation
Standard Deviation (SD) measures how spread out the numbers are from the mean in a dataset.

🧠 Intuition:

Low SD → data points are close to the mean

High SD → data points are more spread out



Where:
σ = standard deviation
xi  = each value
μ = mean
N = number of values

```python
np.std(x)
```
