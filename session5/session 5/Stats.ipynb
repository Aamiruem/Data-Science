# Median


📊 What is Median in Data Science?
Median is a measure of central tendency — it represents the middle value in a sorted list of numbers.

🧠 Definition:
The median is the value that separates the higher half from the lower half of a dataset.

For odd number of elements:
Just pick the middle value after sorting.

Example:
Data: [3, 1, 7]
Sorted: [1, 3, 7]
Median = 3
For even number of elements:
Take the average of the two middle values.

Example:
Data: [4, 1, 6, 2]
Sorted: [1, 2, 4, 6]
Median = (2 + 4) / 2 = 3
🔍 Why is Median Useful?
It isn’t affected by outliers as much as the mean (average).

It's great for skewed data.

Example:
Salaries: [30k, 32k, 31k, 33k, 1 million]
Mean ≈ 225k (misleading)
Median = 32k (more accurate)

🧮 Python Example:
import statistics
data = [3, 1, 7]
print(statistics.median(data))  # Output: 3

```python
np.median(x)
```

# Standard Deviation

📊 Standard Deviation — Short & Clear Explanation
Standard Deviation (SD) measures how spread out the numbers are from the mean in a dataset.

🧠 Intuition:

Low SD → data points are close to the mean

High SD → data points are more spread out



Where:
σ = standard deviation
xi  = each value
μ = mean
N = number of values

```python
np.std(x)
```

```python
import numpy as np  # Import the NumPy library and give it the alias 'np'

x = [1, 2, 3, 4, 5]  # Example list of numbers

result = np.std(x)  # np      → refers to the NumPy library (we imported it as 'np')
# .       → dot notation to access a function from the library
# std     → stands for "standard deviation", a NumPy function
# (x)     → pass the list or array 'x' as the argument
#         → this calculates the standard deviation of values in x

print(result)  # Output the result of the standard deviation
```

```python
np.var(x)
```

🧠 What is Variance?    np.var(x) calculates the variance, which is the average of the squared differences from the mean.

Formula:     Where:

𝑥i is each value

μi is the mean of the values

n is the number of values

🆚 Variance vs Standard Deviation:
np.var(x) → gives variance

np.std(x) → gives the square root of variance

```python
import numpy as np  # import the NumPy library and give it the alias 'np'

x = [1, 2, 3, 4, 5]  # create a list of numbers and store it in variable 'x'

result = np.var(x)  # np      → refers to the NumPy library (aliased as 'np')
# .       → dot operator to access a function inside the library
# var     → stands for "variance", a NumPy function
# (x)     → call the function and pass 'x' as the input data
#         → this calculates the variance of the values in 'x'

print(result)  # print the result to the console
```

```python
print(np.sqrt(np.var(x)))
```

```python
import numpy as np  # Import the NumPy library and give it the alias 'np'

x = [1, 2, 3, 4, 5]  # Define a list 'x' containing numerical values

print(
    np.sqrt(np.var(x))
)  # print(                → built-in Python function to display output
#      np               → refers to the NumPy library (aliased as 'np')
#        .sqrt(         → call the NumPy function 'sqrt' (square root)
#              np       → again, refers to the NumPy library
#                 .var( → call the NumPy function 'var' (variance)
#                      x) → pass the list 'x' to the var() function
#            )          → close the var() function, result is a number (variance)
#   sqrt(...)           → compute square root of that variance
# )                     → close the print() function
```

```python
print(np.std(x, ddof=1)) 
```

```python
import numpy as np  # Import the NumPy library and give it the alias 'np'

x = [1, 2, 3, 4, 5]  # Define a list 'x' containing numerical values

print(
    np.std(x, ddof=1)
)  # print(               → built-in Python function to display the output
#       np             → refers to the NumPy library (aliased as 'np')
#         .std(        → call the NumPy function 'std' (standard deviation)
#              x       → pass the list 'x' as the first argument (data)
#                 ,    → separates multiple arguments in the function
#                  ddof=1  → set the "Delta Degrees of Freedom" to 1
#                            → this calculates the **sample** standard deviation
#                            → by default, ddof=0 (for **population** std)
#            )          → close the std() function call
# )                     → close the print() function


# 🔍 What is ddof=1?
# ddof stands for Delta Degrees of Freedom

# When you set ddof=1, NumPy divides by (n - 1) instead of n

# This is used to calculate the sample standard deviation instead of population standard deviation

# 🧠 Why use ddof=1?
# Use ddof=0 (default) → for entire population (all data points known)

# Use ddof=1 → for sample data (when you're estimating from a smaller group)
```

```python
print(np.var(x, ddof=1))
```
