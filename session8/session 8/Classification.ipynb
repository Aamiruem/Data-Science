```python
# Classification in Data Science
# Classification is a supervised machine learning technique used to predict the category or class of a given data point. It is widely applied in various real-world problems where the output is categorical (e.g., "yes/no", "spam/ham", "disease/no disease").

# 🧠 Core Idea
# You have a dataset with input features (X) and corresponding known outputs or labels (y). A classification model learns the mapping from inputs to outputs, and then can be used to predict the class of new, unseen data.

# 📦 Examples of Classification Problems
# Problem	Input	Output (Class)
# Email filtering	Email text	Spam or Not Spam
# Disease diagnosis	Patient data	Disease or No Disease
# Loan approval	Applicant details	Approved or Rejected
# Image recognition	Image pixels	Cat, Dog, Car, etc.

# 🔧 Popular Classification Algorithms
# Algorithm	Description	Use Case
# Logistic Regression	Simple linear classifier	Binary outcomes
# K-Nearest Neighbors (KNN)	Classifies based on closest training examples	Pattern recognition
# Decision Trees	Tree-like structure of decisions	Interpretability
# Random Forest	Ensemble of decision trees	Robust and powerful
# Support Vector Machine (SVM)	Finds optimal boundary between classes	High-dimensional data
# Naive Bayes	Based on Bayes' theorem	Text classification
# Neural Networks	Deep learning model	Complex patterns, e.g. images, NLP

# 📊 Evaluation Metrics
# Since classification deals with categorical outputs, we use the following metrics:

# Accuracy: (Correct Predictions) / (Total Predictions)

# Precision: TP / (TP + FP)

# Recall (Sensitivity): TP / (TP + FN)

# F1 Score: Harmonic mean of Precision and Recall

# Confusion Matrix: Matrix showing true vs. predicted labels

# 🛠️ Typical Workflow
# Data Collection – Gather labeled data

# Data Preprocessing – Handle missing values, encode categories, normalize

# Train-Test Split – Split dataset into training and testing sets

# Model Selection – Choose appropriate algorithm

# Training – Fit the model on training data

# Evaluation – Test model using metrics

# Tuning – Optimize hyperparameters (e.g. using Grid Search)

# Deployment – Use model to make predictions on real-world data

# 🧪 Example in Python (with Scikit-Learn)
# python
# Copy
# Edit
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Load dataset
X, y = load_iris(return_X_y=True)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)

# Train classifier
model = RandomForestClassifier()
model.fit(X_train, y_train)

# Predict and evaluate
predictions = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, predictions))
```
